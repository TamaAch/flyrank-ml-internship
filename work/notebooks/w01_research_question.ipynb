## 1. My lane (or freestyle) and why

*Name your lane — or say 'freestyle' and describe your own question. One short paragraph: why this one?*

**Lane:** Freestyle: content refresh prioritization.

**Why:** Teams with hundreds of pages cannot refresh all of them, so they need to know which pages to fix first. In the starter sample, 54.2% of pages are labeled declining (trend_direction == "down"), so a random pick would already be right about half the time. Even so, the hand-written baseline rule was right for only 24% of its top 50 picks on the client-holdout test, below that base rate, while a random forest reached 74%. That is roughly 3x the baseline but only about 1.4x the base rate, so I want to understand how much of the model's lift is real and useful. This looks worth studying over the next 7 weeks.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one - typing sentences here breaks Run All.

# Setup: clone the starter repo so the data file exists in this Colab session
import os, subprocess
import pandas as pd

REPO_URL = "https://github.com/flyrank-bih/flyrank-ml-internship-starter"
REPO_DIR = "/content/flyrank-ml-internship-starter"

if not os.path.isdir(REPO_DIR):
    subprocess.run(["git", "clone", "--depth", "1", REPO_URL, REPO_DIR], check=True)
os.chdir(REPO_DIR)  # absolute path, so running twice changes nothing
assert os.path.exists("data/raw/content_refresh_anonymized.csv"), "CSV not found"

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# Base rate: how often a random pick would be a declining page
base_rate = (df["trend_direction"] == "down").mean()
print("Base rate (declining):", round(base_rate, 3))

# Precision@50 values below are copied from outputs/model_report.md (client-holdout split)
baseline_p50 = 0.240
rf_p50 = 0.740
print("Baseline Precision@50:", baseline_p50)
print("Random forest Precision@50:", rf_p50)
print("Lift of random forest vs baseline:", round(rf_p50 / baseline_p50, 2))
print("Lift of random forest vs base rate:", round(rf_p50 / base_rate, 2))
```
